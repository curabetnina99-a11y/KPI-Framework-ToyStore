### 🔍 Understanding the Features

Before modeling, let's understand what we're working with:

**Questions to consider:**
- Which features might be most predictive of survival?

```python
# Check the first few rows
df.head(10)
# Check for missing values
df.isnull().sum()
# Basic statistics
df.describe()
```

**Notice:**

- ⚠️ Some missing values (we're using clean features for now)

```python
from nbresult import ChallengeResult

result = ChallengeResult(
    'data_loading',
    df_shape=df.shape,
    df_columns=df.columns.tolist(),
    survived_exists='Survived' in df.columns
)
result.write()
print(result.check())
```

## Pick X and y

Ok, now that you've visualized the DataFrame, let's create **X** and **y**.


Try to model whether a passenger **Survived** or **not** based on the following features:

- `Pclass`: The class they were in
- `SibSp`: The number of siblings or spouses on the Titanic
- `Parch`: The number of parents or children on the Titanic
- `Fare`: How much they paid for their ticket

```python
X = df[["Pclass", "SibSp", "Parch", "Fare"]]
y = df["Survived"]
```

## Train Test Split

Remember when we said that splitting the data is **one of the most important things in Machine Learning?**

Use Sklearn's `train_test_split` function to split your data into `X_train`, `X_test`, `y_train` and `y_test`.


- Use `random_state=0` to compare your final results with your buddy's results.


```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=0)
```

```python
from nbresult import ChallengeResult

result = ChallengeResult(
    'train_test_split',
    X_train_shape=X_train.shape,
    X_test_shape=X_test.shape,
    y_train_length=len(y_train),
    y_test_length=len(y_test)
)
result.write()
print(result.check())
```

## Scaling

Now we need to scale our features.

Use Sklearn's `Standard Scaler` to bring all of the features to the same scale.

Save your scaled features to two new variables named `X_train_scaled`and `X_test_scaled`.


```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
scaler.fit(X_train)
X_train_scaled = scaler.transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
from nbresult import ChallengeResult
import numpy as np

result = ChallengeResult(
    'scaling',
    X_train_scaled_shape=X_train_scaled.shape,
    X_test_scaled_shape=X_test_scaled.shape,
    X_train_scaled_mean=float(X_train_scaled.mean()),
    X_train_scaled_std=float(X_train_scaled.std())
)
result.write()
print(result.check())
```

## Modelling

Now we can actually run our ML model!

Import, instantiate and fit a **Logistic Regression** model on the training data.

**📝 Important:** Store your fitted model in a variable called `log_model` (this will be used for model validation).


```python
from sklearn.linear_model import LinearRegression
log_model = LinearRegression()
log_model.fit(X_train_scaled, y_train)
```

## Model Evaluation

Score your model on **test data** and store in `accuracy_score`.

Expected: ~`0.71` (71% accuracy)

```python
accuracy_score = log_model.score(X_test_scaled, y_test)
print(f"Score d'évaluation : {accuracy_score:.2f}")
```

How can we interpret this 71% score?

<details>
<summary>Answer 💡</summary>

The default scoring metric for a **Logistic Regression** model is called the `accuracy`. \
It basically means that when running predictions on our test set, 71% of the returned values would be correctly predicted. You will get more information on the scoring metrics on future lectures.

**71% accuracy** means the model correctly predicts 71% of passengers.

71% is pretty good given we only used 4 features!

</details>

## Prediction


We've found the records of a passenger that wasn't in the original dataset!

Use the model we've just created to predict whether they would have survived or not!

**📝 Important:**
- Store the new passenger data in a variable called `new_record` (already provided below)
- Scale it and store as `new_record_scaled`
- Store the prediction in `new_prediction`


<details>
    <summary>Hint 🔍</summary>
    
Don't forget to scale the data, but don't fit your scaler on this new data, just transform it (otherwise, you're causing **data leakage**).
    
     
</details>

```python
import pandas as pd

# 1. Définition du nouveau passager avec les colonnes demandées
new_record = pd.DataFrame(
    {
        "Pclass": [3],  # 3ème classe
        "SibSp": [0],  # Nombre de frères, sœurs / conjoint à bord
        "Parch": [0],  # Nombre de parents / enfants à bord
        "Fare": [7.25],  # Prix du billet
    }
)

# 2. Mise à l'échelle avec le scaler déjà entraîné
new_record_scaled = scaler.transform(new_record)

# 3. Prédiction avec le modèle
new_prediction = log_model.predict(new_record_scaled)

# 4. Affichage du résultat
print(f"Prediction : {new_prediction[0]}")
```

How do you interpret the result?


> **The prediction is class 1, indicating that the passenger would probably have survived!**

> Ajouter une citation




```python

from nbresult import ChallengeResult

result = ChallengeResult(
    'model',
    model_type=type(log_model).__name__,
    model_score=float(log_model.score(X_test_scaled, y_test)),
    new_prediction=int(log_model.predict(new_record_scaled)[0])
)
result.write()
print(result.check())
```
